```python
from catboost import CatBoostClassifier
train['Timestamp'] = pd.to_datetime(train['Timestamp'])
model = CatBoostClassifier()
model.fit(train.drop(columns = ['Attack Type', 'ID']), train['Attack Type'])
```

```python
test = pd.read_csv("/content/test_data.csv")
test['Timestamp'] = pd.to_datetime(test['Timestamp'])
```

```python
predictions = model.predict(test.drop(columns = ['ID']))
print(len(predictions), len(test['ID']))
```

```python
submission2 = pd.DataFrame({
    'subtaskID':2,
    'datapointID': test['ID'],
    'answer':predictions.flatten(),
})
```

```python

submission = pd.concat([submission1, submission2], ignore_index = True)
submission.to_csv('submission.csv', index = False)
```
